### Dictionary

```python
## Creating Dictionaries
empty_dict={}
print(type(empty_dict))
```

```python
empty_dict=dict()
empty_dict
```

```python
student = {"name" : "Priyanshu", "age" : 21, "grade" : 24}
print(student)
print(type(student))
```

```python
## Single key always used
student = {"name" : "Priyanshu", "age" : 21, "name" : 24}
print(student)
```

```python
## accessing dictionay elements
student = {"name" : "Priyanshu", "age" : 21, "grade" : 'A'}
print(student['grade'])

## Accessing using get() method
print(student.get('grade'))
print(student.get('last_name'))
print(student.get('last_name', "Not Available"))
```

```python
## Modifying Dictionary elements
## Dictionary are mutable, so you can add, update or delete elements
print(student)
```

```python
student['age'] = 18 ## Updated value for the key
print(student)
student["address"] = "India" ## added a new key value
print(student)
```

```python
del student['grade'] ## delete key and value pair

print(student)
```

```python
## Dictinary methods

keys = student.keys() ##get all the keys
print(keys)
values = student.values() ## get all the values
print(values)

items = student.items() ## get all keys value pairs
print(items)
```

```python
## Shallow copy
student_copy = student
print(student)
print(student_copy)
student["name"] = "Sonu"
```

```python
student["name"] = "Sonu"
print(student)
print(student_copy)
```

```python
student_copy1 = student.copy() ## Shallow copy
print(student_copy1)
print(student)
```

```python
student["name"] = "Priyanshu"
print(student_copy1)
print(student)
```

```python
## Iterating Over dictionary
## You can use loops to iterate over dictionary, keys, values, or items

## Iterating over keys
for keys in student.keys():
    print(keys)
```

```python
## Iterate over values
for value in student.values():
    print(value)
```

```python
## Iterate over key value pairs
for key, value in student.items():
    print(f"{key}:{value}")
```

```python
## Nested Dictionaries
students = {
    "student1" : {"name" : "Rohit", "age" : 21},
    "student2" : {"name" : "Ariz", "age" : 25}
}
print(students)
```

```python
## Access Nested dictionary
print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
## Iterating Over nested dictionary
for student_id, student_info in students.items():
    print(f"{student_id} : {student_info}")
    for key, value in student_info.items():
        print(f"{key}:{value}")
```

```python
## Dictionary Comprehension
squares = {x : x**2 for x in range(5)}
print(squares)
```

```python
## Conditional dictionary Comprehension
evens = { x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
## Practical Examples

## Use a dictionary to count the frequency of elements in list

numbers = [1,2,3,3,3,4,4,4,5,5,7,8,8]
frequency = {}

for number in numbers:
    if number in frequency:
        frequency[number] += 1
    else:
        frequency[number] = 1
        
print(frequency)
```

```python
## Merge 2 dictionary into one

dict1 = {"a":1, "b":2}
dict2={"b":3,"c":4}
merged_dict = {**dict1, **dict2}
print(merged_dict)
```
