Sets

```python
## Create a set
my_set={1, 2, 3, 4, 5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set=set()
print(type(my_empty_set))
```

```python
my_set=set([1,2,3,4,5])
print(my_set)
```

```python
my_set=set([1,2,3,4,3,4,5])
print(my_set)
```

```python
## Basics Sets Operation
## Adding and Removing elements
my_set.add(7)
print(my_set)
```

```python
## Remove the elements from a set
my_set.remove(3)
print(my_set)
```

```python
my_set.remove(10)
```

```python
my_set.discard(2)
print(my_set)
```

```python
my_set.discard(11)
print(my_set)
```

```python
## Pop method
removed_elements=my_set.pop()
print(removed_elements)
print(my_set)
```

```python
## Clear all the elements
my_set.clear()
print(my_set)
```

```python
## Set Membership test
my_set = {1,2,3,4,5}
print(3 in my_set)
print(10 in my_set)
```

```python
## Mathematical Operation
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

## Union
union_set = set1.union(set2)
print(union_set)

## Intersection
intersection_set = set1.intersection(set2)
print(intersection_set)

set1.intersection_update(set2)
print(set1)
```

```python
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

## Difference
print(set1.difference(set2))
```

```python
## Symmetric difference
set1.symmetric_difference(set2)
```

```python
## Sets Methods
set1 = {1,2,3, 4, 5}
set2 = {3,4,5}

## is subset
print(set1.issubset(set2))

set1.issuperset(set2)
```

```python
lst=[1,2,2,3,4,5]
set(lst)
```

```python
## Counting unique words in text

text = "In this tutorial we are discussing about sets"

words = text.split()

## Convert list of words to set to get unique words
unique_words=set(words)
print(unique_words)
print(len(unique_words))
```
