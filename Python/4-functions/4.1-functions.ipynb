### Functions in Python

A function is a block of code that performs specific task.help in organising code, reusing code, and improving readability.

```python
## Syntax
def function_name(parameter):
    """"Docstring"""
    # Function body
    return expression
```

```python
## Why functions?
num = 24
if num % 2 == 0:
    print("Number is even.")
else:
    print("Number is odd.")
```

```python
def even_or_odd(num):
    """"This function perform finds even or odd"""
    if num % 2 == 0:
        print("The number is even.")
    else:
        print("The number is odd")
```

```python
## Call this function
even_or_odd(24)
```

```python
## Function with multiple parameters

def add(a, b):
    return a + b

result = add(2, 4)
print(result)
```

```python
## Default Parameters
def greet(name="Guest"):
    print(f"Hello {name}, welcome again !")

greet("Krish")
greet()
```

```python
### Variable length Arguments
## Positional and Keywords Arguments

## Positional Arguments


def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6,7,8, "Sonu")
```

```python
## Keywords Arguments  key-value pairs
def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="Priyanshu Kumar", age="21", country="India")
```

```python
## Keywords Arguments  key-value pairs
def print_details(*args,**kwargs):
    
    for val in args:
        print(f"Positional Argument: {val}")
        
    for key, value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,4,5,6, "Sonu",name="Priyanshu Kumar", age="21", country="India")
```

```python
### Return Statements
def multiply(a, b):
    return a*b

multiply(2, 3)
```

```python
### Return Statements
def multiply(a, b):
    return a*b, a

multiply(2, 3)
```
