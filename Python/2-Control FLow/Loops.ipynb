### While Loop

```python
## While Loop
## the while loop continues to executes as long as the condition is true.

count = 0

while count < 5:
    print(count)
    count += 1
```

```python
count = 0
while count%2==0:
    print(count)
    count += 1
```

```python
## Loop Control Statements

## break
## The break statement exists the loop permaturely

for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
## continue
## THe continue statement skips the current iteration and continues with the next.

for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
## pass
## The pass statement is a null operation; it does nothing.

for i in range(5):
    if i == 3:
        pass
        print("Number is pass")
    print(i)
```

```python
## Nested loop
## a loop inside a loop

for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
## Examples - Calculate the sum of first N natural number using a while and for loop

## while loop

n = 10
sum = 0
count = 1

while count <= n:
    sum += count
    count += 1
    
print(f"Sum of first {n} natural number:",sum)
```

```python
n = 10
sum = 0
count = 1
for i in range(11):
    sum += i
    
print(sum)
```

```python
## Example - Prime number between 1 to 100

for num in range(1, 101):
    if num>1:
        for i in range(2, num):
            if num%i == 0:
                break
        else:
            print(num)
```
